```python
import json
import os
from pathlib import Path
import shutil
import sys

DATASET = Path('/kaggle/input/datasets/evanpieser/qcompile-verifiable')
STARTER_KIT = DATASET / 'starter_kit'
RUN_DIR = Path('/kaggle/working/dfthcsos_run')
KICKMIX = Path('/kaggle/input/competitions/google-verifiable-quantum-circuit-compilation/kickmix')
sys.path.insert(0, str(KICKMIX))
sys.path.insert(0, str(STARTER_KIT))
sys.path.insert(0, str(DATASET))
import kickmix
print('kickmix:', kickmix.__file__)
print('CircuitBuilder available:', hasattr(kickmix, 'CircuitBuilder'))
print('Dataset:', DATASET)
print('Run directory:', RUN_DIR)
```

```python
from starter.task3_circuits.build_circuits import build_submission, format_summary
from starter.task1_factorization.solution_io import BASELINES_DIRECTORY
from verifier.reference import load_systems

systems = load_systems()
factorization = BASELINES_DIRECTORY / systems['FeMoCo76o113e']['baseline']
print('Factorization:', factorization)
summary = build_submission(
    'FeMoCo76o113e',
    factorization,
    output_directory=RUN_DIR,
    overwrite=True,
)
print(format_summary(summary))
```

```python
from verifier.verify import verify

verification = verify('FeMoCo76o113e', RUN_DIR)
(RUN_DIR / 'verification.json').write_text(json.dumps(verification, indent=2, default=float) + '\n')
print(json.dumps(verification, indent=2, default=float))
shutil.make_archive('/kaggle/working/dfthcsos_submission', 'zip', RUN_DIR)
print('Packaged:', '/kaggle/working/dfthcsos_submission.zip')
```
