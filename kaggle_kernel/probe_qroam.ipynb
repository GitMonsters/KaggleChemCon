```python
import inspect
import sys
from pathlib import Path
import numpy as np
sys.path.insert(0, '/kaggle/input/competitions/google-verifiable-quantum-circuit-compilation/kickmix')
sys.path.insert(0, '/kaggle/input/datasets/evanpieser/qcompile-verifiable/starter_kit')
from qualtran.bloqs.data_loading import QROM, SelectSwapQROM
print('SelectSwapQROM:', inspect.signature(SelectSwapQROM))
print('build_from_data:', inspect.signature(SelectSwapQROM.build_from_data))
from dfthc.build_angle_k import AngleK
from starter.task3_circuits.angle_k_kickmix import angle_k_to_circuit
delta = np.arange(128, dtype=np.int64).reshape(32, 4) % 4096
for block in (None, 1):
    if block is None:
        qrom = QROM.build_from_data(delta, target_bitsizes=(12,), target_shapes=((4,),))
    else:
        qrom = SelectSwapQROM.build_from_data(
            delta, target_bitsizes=(12,), log_block_sizes=(block,)
        )
    circuit = angle_k_to_circuit(AngleK(k=0, qrom=qrom))
    print('block', block, 'qubits', circuit.num_qubits, 'stats', circuit.operation_statistics())
```
